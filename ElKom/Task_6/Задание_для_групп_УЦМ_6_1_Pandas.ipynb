> Имеется *серия*, содержащая числа от 10 до 1000. Какое число содержит элемент с индексом 175? Вычислите разность элементов с индексами 900 и 16.

**Серия** — это последовательность всех целых чисел от 10 до 1000 включительно, где элемент с индексом 0 равен 10, с индексом 1 — 11 и т.д., то:

Элемент с индексом $k$: $a_k = 10 + k$.

При $k = 175$: $a_{175} = 10 + 175 = 185$.

$a_{900} = 10 + 900 = 910$.

$a_{16} = 10 + 16 = 26$.

Разность: $910 - 26 = 884$.

### Как получить доступ к элементам в pandas (`iloc` и `loc`)

**`iloc`** и **`loc`** — это индексаторы для доступа к строкам и столбцам в таблицах (`DataFrame`) и сериях (`Series`) библиотеки **pandas**. Они нужны, чтобы точно выбирать, фильтровать или изменять нужные данные, не путаясь в номерах строк и названиях индексов.

**Базовый синтаксис:** `df.loc[строки, столбцы]` или `df.iloc[строки, столбцы]`

| Критерий | `iloc` (*integer location*) | `loc` (*location*) |
| :--- | :--- | :--- |
| **Что принимает** | Целочисленные позиции (`0, 1, 2, …`) | Метки индекса (*labels*) |
| **По чему работает** | По позиции в массиве | По значению индекса |
| **Срезы** | Не включают правую границу (`0:2` → 0 и 1) | Включают правую границу (`'a':'c'` → a, b, c) |
| **Тип индекса** | Не важен | Важен |

#### Пример на практике
```python
import pandas as pd

df = pd.DataFrame({'Имя': ['Аня', 'Боря', 'Вера']}, index=['id1', 'id2', 'id3'])

# По позиции (iloc):
df.iloc[0]          # Первая строка ('Аня')
df.iloc[0:2, 0]     # Строки 0 и 1, первый столбец ('Аня', 'Боря')

# По метке (loc):
df.loc['id1']       # Строка с индексом 'id1' ('Аня')
df.loc['id1':'id2', 'Имя'] # Строки от 'id1' до 'id2' ВКЛЮЧИТЕЛЬНО ('Аня', 'Боря')
```

### Как создать `DataFrame` в pandas

**`DataFrame`** — это основная структура данных в библиотеке **pandas**, представляющая собой двумерную таблицу со строками и столбцами (как лист в Excel или таблица в SQL). Она нужна для удобного хранения, очистки и анализа структурированных данных.

**Базовый синтаксис:** `pd.DataFrame(data, index=..., columns=...)`

| Способ создания | Что подаётся на вход | Когда удобно использовать |
| :--- | :--- | :--- |
| **Из словаря списков** | `{'Столбец': [значения]}` | Самый частый способ для ручного создания |
| **Из списка словарей** | `[{'col1': 1, 'col2': 2}, ...]` | Когда данные приходят построчно (например, из JSON) |
| **Из списка списков** | `[[1, 2], [3, 4]]` + `columns=[...]` | Когда есть готовая матрица или двумерный массив |
| **Из внешнего файла** | `pd.read_csv('file.csv')` | Для загрузки реальных датасетов (CSV, Excel, SQL) |

#### Примеры на практике

### Теоретический минимум. Как разделить столбец на несколько в pandas (`.str.split`)

**`.str.split()`** — это строковый метод в **pandas**, который разбивает текст в каждой ячейке столбца по заданному разделителю. В паре с параметром `expand=True` он позволяет быстро разнести данные из одного столбца сразу по нескольким новым.

**Базовый синтаксис:** `df[['col1', 'col2']] = df['col'].str.split('разделитель', n=1, expand=True)`

| Параметр | Что принимает | Зачем нужен |
| :--- | :--- | :--- |
| **`pat`** (1-й аргумент) | Строку (`' '`, `','`, `'-'`) | Символ-разделитель, по которому режется текст (по умолчанию пробел) |
| **`expand=True`** | `True` или `False` | Превращает результат не в списки внутри одной ячейки, а сразу в новые столбцы |
| **`n`** | Целое число (`1`, `2`, …) | Ограничивает число делений (защищает от ошибки, если в строке лишние пробелы) |

#### Пример на практике
```python
import pandas as pd

market = pd.DataFrame({
    'car': ['Lada Granta', 'Kia Rio', 'Land Rover Range Rover'],
    'year': [2019, 2010, 2021]
})

# Делим по первому пробелу (n=1) и сразу записываем в два новых столбца
market[['mark', 'model']] = market['car'].str.split(' ', n=1, expand=True)

# Удаляем исходный столбец, если он больше не нужен
market = market.drop(columns=['car'])
```

### Описательная статистика в pandas (`.min()`, `.mean()`, `.std()` и др.)

**Статистические (агрегирующие) методы** в **pandas** позволяют быстро свести целый столбец чисел к одному итоговому показателю. Они автоматически пропускают пустые значения (`NaN`) и нужны для первичного анализа данных.

**Базовый синтаксис:** `df['столбец'].метод()` или `df.describe()` (для сводки сразу по всем числам)

| Метод | Что вычисляет | Когда полезен |
| :--- | :--- | :--- |
| **`.min()` / `.max()`** | Минимум и максимум | Найти границы диапазона значений или выбросы |
| **`.mean()`** | Среднее арифметическое | Оценить средний уровень, когда нет сильных выбросов |
| **`.median()`** | Медиану (середину ряда) | Оценить типичное значение, устойчивое к резким скачкам |
| **`.std()`** | Стандартное отклонение | Понять разброс (волатильность) данных вокруг среднего |

#### Пример на практике
```python
import pandas as pd

df = pd.DataFrame({
    'Open':,
    'High':,
    'Low':,
    'Close': [104, 103, 106, 112]
})

min_close = df['Close'].min()       # 103 (минимум закрытия)
max_close = df['Close'].max()       # 112 (максимум закрытия)
std_open = df['Open'].std()         # ~4.35 (разброс цены открытия)
mean_low = df['Low'].mean()         # 101.5 (средний минимум)
median_high = df['High'].median()   # 107.5 (медиана максимума)
```
