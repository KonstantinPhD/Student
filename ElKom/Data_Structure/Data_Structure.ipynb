# Структуры данных в Python

## list — список
Упорядоченная изменяемая коллекция. Быстрый доступ по индексу, но медленная вставка/удаление в начало.
```python
nums = [1, 2, 3]
nums.append(4)
```

## tuple — кортеж
Упорядоченная **неизменяемая** последовательность. Быстрее списка, может быть ключом словаря.
```python
point = (10, 20)
```

## set — множество
Неупорядоченная коллекция **уникальных** элементов. Быстрая проверка вхождения (O(1)).
```python
tags = {"python", "code", "python"}  # {"python", "code"}
```

## frozenset — неизменяемое множество
Как `set`, но **неизменяемый**. Можно использовать как ключ словаря.
```python
fs = frozenset([1, 2, 3])
```

## dict — словарь
Коллекция пар **ключ → значение**. Быстрый поиск по ключу (O(1)). С Python 3.7 сохраняет порядок вставки.
```python
user = {"name": "Anna", "age": 25}
```

# Структуры данных в Python

## list (список)

**Для чего:** упорядоченная коллекция изменяемых элементов. Подходит, когда нужен порядок и возможность менять содержимое.

**Создание:**
```python
lst = [1, 2, 3]
lst = list((1, 2, 3))
lst = [i for i in range(5)]   # [0, 1, 2, 3, 4]
```

**Основные операции:**
```python
lst.append(4)        # добавить в конец
lst.insert(0, 0)     # вставить по индексу
lst.remove(2)        # удалить по значению
lst.pop()            # удалить последний
lst[0]               # доступ по индексу
lst[1:3]             # срез
```

---

## tuple (кортеж)

**Для чего:** неизменяемая последовательность. Используется для данных, которые не должны меняться, и как ключи словаря.

**Создание:**
```python
t = (1, 2, 3)
t = 1, 2, 3
t = tuple([1, 2, 3])
t = (5,)             # кортеж из одного элемента (запятая обязательна)
```

**Основные операции:**
```python
t[0]                 # доступ по индексу
t[1:3]               # срез
len(t)               # длина
t.count(2)           # количество вхождений
t.index(3)           # индекс элемента
```

---

## set (множество)

**Для чего:** хранение уникальных элементов, быстрая проверка принадлежности, математические операции (объединение, пересечение).

**Создание:**
```python
s = {1, 2, 3}
s = set([1, 2, 2, 3])   # {1, 2, 3}
s = {i for i in range(5)}
empty = set()           # НЕ {}, это пустой словарь
```

**Основные операции:**
```python
s.add(4)             # добавить элемент
s.remove(2)          # удалить (ошибка, если нет)
s.discard(10)        # удалить без ошибки
s.union({5, 6})      # объединение
s.intersection({2, 3})  # пересечение
s.difference({1})    # разность
2 in s               # проверка вхождения — O(1)
```

---

## frozenset

**Для чего:** неизменяемое множество. Можно использовать как ключ словаря или элемент другого множества.

**Создание:**
```python
fs = frozenset([1, 2, 3])
fs = frozenset({1, 2, 3})
```

**Основные операции:**
```python
fs.union({4})        # возвращает новый frozenset
fs.intersection({1})
1 in fs
# fs.add(4)          # ошибка — неизменяемо
```

## dict (словарь)

**Для чего:** хранение пар «ключ → значение», быстрый поиск по ключу.

**Создание:**
```python
d = {"a": 1, "b": 2}
d = dict(a=1, b=2)
d = dict([("a", 1), ("b", 2)])
d = {k: k**2 for k in range(3)}   # {0: 0, 1: 1, 2: 4}
```

**Основные операции:**
```python
d["a"]               # доступ по ключу
d["c"] = 3           # добавить / изменить
d.get("x", 0)        # безопасный доступ
d.keys()             # ключи
d.values()           # значения
d.items()            # пары
d.pop("a")           # удалить по ключу
d.update({"e": 5})   # объединить
```

## str (строка)

**Для чего:** неизменяемая последовательность символов для текста.

**Создание:**
```python
s = "hello"
s = 'hello'
s = """многострочная
строка"""
s = str(123)
```

**Основные операции:**
```python
s.upper()            # ВЕРХНИЙ РЕГИСТР
s.lower()            # нижний регистр
s.split(",")         # разбить на список
",".join(["a", "b"]) # склеить
s.replace("h", "H")  # замена
s.strip()            # убрать пробелы по краям
s[0]                 # символ по индексу
s[::-1]              # разворот
```

## bytes

**Для чего:** неизменяемая последовательность байт (0–255). Для бинарных данных, сети, файлов.

**Создание:**
```python
b = b"hello"
b = bytes([104, 101, 108, 108, 111])
b = "hello".encode("utf-8")
```

**Основные операции:**
```python
b[0]                 # 104
b.decode("utf-8")    # обратно в str
b.hex()              # hex-представление
```

---

## bytearray

**Для чего:** изменяемая версия bytes. Удобно, когда нужно менять байты на месте.

**Создание:**
```python
ba = bytearray(b"hello")
ba = bytearray([1, 2, 3])
```

**Основные операции:**
```python
ba[0] = 72           # изменение
ba.append(33)
ba.extend(b"!!")
ba.decode("utf-8")
```

## collections.deque

**Для чего:** двусторонняя очередь с быстрыми операциями добавления/удаления с обоих концов. Идеальна для стеков и очередей.

**Создание:**
```python
from collections import deque

dq = deque([1, 2, 3])
dq = deque(maxlen=5)   # ограниченный размер
```

**Основные операции:**
```python
dq.append(4)         # в конец
dq.appendleft(0)     # в начало
dq.pop()             # с конца
dq.popleft()         # с начала
dq.rotate(1)         # сдвиг
dq.extend([5, 6])
```

## collections.Counter

**Для чего:** подсчёт количества элементов. Удобно для частотного анализа.

**Создание:**
```python
from collections import Counter

c = Counter("abracadabra")
c = Counter([1, 1, 2, 3, 3, 3])
c = Counter(a=2, b=1)
```

**Основные операции:**
```python
c["a"]                       # сколько раз встречается
c.most_common(2)             # топ-2 частых
c.update("aaa")              # добавить элементы
c.elements()                 # развернуть обратно
c1 + c2                      # сложение счётчиков
```

## collections.defaultdict

**Для чего:** словарь, который автоматически создаёт значение для отсутствующего ключа.

**Создание:**
```python
from collections import defaultdict

dd = defaultdict(int)        # значение по умолчанию 0
dd = defaultdict(list)       # значение по умолчанию []
dd = defaultdict(lambda: "?")
```

**Основные операции:**
```python
dd["a"] += 1                 # не вызовет KeyError
dd["x"].append(1)
dd["missing"]                # вернёт значение по умолчанию
```

---

## collections.OrderedDict

**Для чего:** словарь с гарантированным порядком вставки (актуально до Python 3.7, сейчас обычный dict тоже упорядочен).

**Создание:**
```python
from collections import OrderedDict

od = OrderedDict()
od = OrderedDict([("a", 1), ("b", 2)])
```

**Основные операции:**
```python
od["c"] = 3
od.move_to_end("a")          # переместить ключ в конец
od.popitem(last=False)       # удалить первый
od.move_to_end("c", last=False)
```

---

## collections.namedtuple

**Для чего:** кортеж с именованными полями. Лёгкая альтернатива классу для простых записей.

**Создание:**
```python
from collections import namedtuple

Point = namedtuple("Point", ["x", "y"])
p = Point(1, 2)
p = Point(x=1, y=2)
```

**Основные операции:**
```python
p.x                  # доступ по имени
p[0]                 # доступ по индексу
p._replace(x=10)     # новый экземпляр с изменением
p._asdict()          # превратить в dict
```

---

## array.array

**Для чего:** компактный массив однотипных чисел. Экономит память по сравнению со списком.

**Создание:**
```python
from array import array

a = array("i", [1, 2, 3])    # int
a = array("f", [1.0, 2.5])   # float
a = array("d", [1.0, 2.5])   # double
```

**Основные операции:**
```python
a.append(4)
a.insert(0, 0)
a[0]                 # доступ
a.tolist()           # в список
a.tobytes()          # в байты
```

## Шпаргалка: что выбрать?

| Задача | Структура |
|--------|-----------|
| Упорядоченный изменяемый набор | `list` |
| Неизменяемые данные | `tuple` |
| Уникальные элементы | `set` |
| Уникальные + неизменяемые | `frozenset` |
| Пары ключ-значение | `dict` |
| Очередь / стек | `deque` |
| Подсчёт частот | `Counter` |
| Словарь с дефолтом | `defaultdict` |
| Запись с полями | `namedtuple` |
| Компактный числовой массив | `array` |
