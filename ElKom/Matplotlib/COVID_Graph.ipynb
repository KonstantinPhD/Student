```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt;
```

```python
df = pd.read_csv('01_data03.csv');
df.index = pd.Index(df.date);
df.head()
```

```python
df['Date'] = pd.to_datetime(df['date'])

plt.figure(figsize=(5,3))
plt.plot(df.Date,df.RUS, label='Russia')
#plt.plot(df.Date,df.USA, label='USA')
plt.plot(df.Date,df.CHN, label='CHN')
plt.xticks(rotation=45)  # Поворот подписей дат для лучшей читаемости
plt.tight_layout()
plt.legend()  
plt.grid()        # ← displays the legend
plt.show()
```
