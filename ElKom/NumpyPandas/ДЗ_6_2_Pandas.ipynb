# Шпаргалка команд для ДЗ 6.2 (Pandas & NumPy)

## 1. Загрузка и первичный осмотр (Task 1)
* `pd.read_csv('path/to/file.csv')` — чтение CSV-файла в `DataFrame`.
* `df.head(n)` / `df.tail(n)` — вывод первых или последних `n` строк таблицы.
* `df.shape` — размерность таблицы `(строки, столбцы)`.
* `df.info()` — сводка по столбцам: типы данных и количество непустых значений.
* `df.describe()` — основные статистики (среднее, стандартное отклонение, минимум, максимум, квартили).
* `df.isna().sum()` — подсчёт количества пропусков (`NaN`) в каждом столбце.
* `df['col'].unique()` — массив уникальных значений в столбце.
* `df['col'].value_counts()` — частота встречаемости каждого уникального значения.

## 2. Фильтрация, преобразование и сводные таблицы (Task 2)
* `df[df['col'] == 'val']` — фильтрация строк по точному совпадению.
* `df[df['col'].isin(['val1', 'val2'])]` — фильтрация строк по списку допустимых значений.
* `df.loc[условие, 'col']` — доступ к конкретным строкам и столбцам по меткам (для пересчёта единиц измерения).
* `pd.to_datetime(df['date_col'])` — преобразование строк в формат даты и времени (`datetime64`).
* `df.pivot_table(index=..., columns=..., values=..., aggfunc=...)` — создание сводной таблицы (перевод длинного формата данных в широкий).
* `df.sort_index(ascending=False)` — сортировка таблицы по индексу (датам) по убыванию.
* `df.rename(columns={'old': 'new'})` — переименование столбцов.

## 3. Работа с временными рядами и интерполяция (Task 3)
* `s.dropna()` — удаление пропущенных значений из серии или столбца.
* `s.resample('QE').asfreq()` — изменение частоты временного ряда (`'QE'` — конец квартала, `'ME'` — конец месяца, `'YE'` — конец года) с созданием пустых строк для промежуточных дат.
* `pd.date_range(start=..., end=..., periods=...)` — генерация равномерной сетки дат.
* `s.reindex(new_index)` — переиндексация ряда под новую сетку дат (новые точки заполняются `NaN`).
* `s.interpolate(method=...)` — заполнение пропусков интерполяцией в `pandas` (методы: `'linear'`, `'time'`, `'spline'`, `'polynomial'`).
* `np.interp(x_new, x_known, y_known)` — одномерная линейная интерполяция в `numpy` (работает с числовыми массивами).

## 4. Скользящие и кумулятивные окна (Task 4)
* `s.expanding().mean()` — кумулятивное (накопительное) среднее от начала ряда до текущей точки.
* `s.expanding().var()` — кумулятивная дисперсия.
* `s.rolling(window=k).mean()` — скользящее среднее по окну размером `k`.
* `s.rolling(window=k).var()` — скользящая дисперсия по окну размером `k`.
* `s.rolling(window=k).std()` — скользящее стандартное отклонение.

## 5. Анализ прироста, тренды и прогноз (Task 5)
* `pd.DateOffset(years=10)` — смещение даты на заданный интервал (для отбора последних 10 лет).
* `df.iloc[0]` / `df.iloc[-1]` — обращение к первой и последней строке по числовой позиции.
* `s.pct_change()` — процентное изменение между соседними элементами.
* `np.arange(start, stop)` — создание массива последовательных целых чисел (для оси времени X).
* `np.polyfit(x, y, deg=1)` — подбор коэффициентов полинома степени `deg` методом наименьших квадратов (`deg=1` — линейный тренд).
* `np.poly1d(coeffs)` — создание функции полинома из коэффициентов для расчёта значений в новых точках.
* `np.std(residuals)` — вычисление стандартного отклонения (для оценки разброса ошибок и границ неопределённости).

## 6. Визуализация (`matplotlib.pyplot`)
* `plt.figure(figsize=(w, h))` — создание полотна заданного размера.
* `plt.plot(x, y, label=...)` — построение линейного графика.
* `plt.scatter(x, y, label=...)` — построение точечного графика (для выделения исходных или новых точек).
* `plt.fill_between(x, y_min, y_max, alpha=0.2)` — заливка области между двумя кривыми (для отображения доверительного интервала / неопределённости).
* `plt.title(...)` / `plt.xlabel(...)` / `plt.ylabel(...)` — добавление заголовка и подписей осей.
* `plt.legend()` / `plt.grid(True)` — отображение легенды и координатной сетки.
* `plt.show()` — вывод графика на экран.
