```python
import numpy as np

np.array([1, 2, 3])                 # Матрица из списка
np.zeros((2, 3))                    # Матрица 2x3 из нулей
np.ones((3, 3))                     # Матрица 3x3 из единиц
np.full((2, 2), 7)                  # Матрица 2x2, заполненная семерками
np.eye(4)                           # Единичная матрица 4x4
np.arange(0, 10, 2)                 # От 0 до 10 (не вкл.) с шагом 2: [0, 2, 4, 6, 8]
np.linspace(0, 1, 5)                # 5 чисел от 0 до 1 включительно

np.random.seed(42)                  # Фиксация генератора (для повторяемости)
np.random.rand(2, 3)                # Вещественные числа от 0 до 1
np.random.randint(1, 10, (2, 2))    # Целые числа от 1 до 9 (10 не вкл.), размер 2x2
np.random.choice([1, 5, 9], 3)      # Случайный выбор 3 элементов из списка

arr[0, 1]                           # Элемент в 0-й строке, 1-м столбце
arr[0]                              # Вся первая строка
arr[:, 1]                           # Весь второй столбец
arr[1:3, 0:2]                       # Срез: строки 1-2, столбцы 0-1
arr[-1]                             # Последняя строка

mask = arr > 5                      # Массив из True/False
arr[mask]                           # Только элементы больше 5 (одномерный массив)
arr[arr % 2 == 0]                   # Только четные элементы
np.where(arr > 5, arr, 0)           # Где > 5 — оставить число, иначе 0

np.maximum(ar1, ar2)                # Поэлементный максимум из двух массивов
np.minimum(ar1, ar2)                # Поэлементный минимум
arr + 5                             # Прибавить 5 к каждому элементу
arr * ar2                           # Поэлементное умножение (не матричное!)
np.dot(ar1, ar2)                    # Матричное умножение (или оператор @)

arr.sum()                           # Сумма всех элементов
arr.mean()                          # Среднее арифметическое
arr.max(), arr.min()                # Максимум и минимум
arr.sum(axis=0)                     # Сумма по столбцам
arr.sum(axis=1)                     # Сумма по строкам  

np.unique(arr)                      # Уникальные значения (отсортированные)
np.sort(arr)                        # Сортировка (возвращает копию)
np.argmax(arr), np.argmin(arr)      # Индекс максимума / минимума
np.argwhere(arr == 10)              # Координаты (индексы) всех элементов = 10
np.deg2rad(90)                      # Градусы -> Радианы
np.rad2deg(np.pi)                   # Радианы -> Градусы
